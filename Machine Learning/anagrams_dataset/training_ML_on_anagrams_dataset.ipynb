```python
from sklearn.tree import DecisionTreeClassifier
import pandas as pd
```

```python
dataset_train = pd.read_excel("anagrams_dataset_train.xlsx")
dataset_test = pd.read_excel("anagrams_datasets_test.xlsx")

DTC = DecisionTreeClassifier()
x_train = dataset_train.iloc[:,2:]   #num1 num2 num3
y_train = dataset_train.iloc[:,1:-3]  #labeled data
DTC.fit(x_train.values,y_train.values)
predict_train = DTC.predict([[3,5.5,9]])
predict_train
```

```python
x_test = dataset_test.iloc[:,2:] 
predict_test = DTC.predict(x_test.values)
predict_test
```

```python
total_data = dataset_test.iloc[:,1:-3]
actual = total_data.values
count = 0
for i in range(0,len(actual)):
    if actual[i] == predict_test[i]:
        count += 1
print((count * 100)/len(predict_test))
```
