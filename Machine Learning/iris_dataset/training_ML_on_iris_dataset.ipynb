```python
import pandas as pd 
from sklearn.ensemble import RandomForestClassifier

data = pd.read_excel("iris_dataset_train.xlsx")
data_test = pd.read_excel("iris_dataset_test.xlsx")
```

```python
x = data.iloc[:,:-1]
y = data.iloc[:,-1]
```

```python
model = RandomForestClassifier()
model.fit(x.values,y.values)
predict = model.predict([[5.7,4.4,1.5,0.4]])
predict
```

```python
x_test = data_test.iloc[:,:-1]
y_test = data_test.iloc[:,-1]
#model.fit(x_test.values,y_test.values)
predict = model.predict(x_test.values)
predict
```

```python
actual = data_test.iloc[:,-1]
count = 0
for i in range(0,len(actual)):
    if actual[i] == predict[i]:
        count += 1
```

```python
print((count *100)/len(predict))
```
