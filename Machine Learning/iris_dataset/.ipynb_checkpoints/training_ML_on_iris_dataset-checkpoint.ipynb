```python
import pandas as pd 
from sklearn.ensemble import RandomForestClassifier

data = pd.read_excel("iris_dataset_train.xlsx")
data_test = pd.read_excel("iris_dataset_test.xlsx")
```

```python
x = data.iloc[:,:-1]
y = data.iloc[:,-1]
```

```python
model = RandomForestClassifier()
model.fit(x.values,y.values)
predict = model.predict([[5.7,4.4,1.5,0.4]])
predict
```
