```python
from sklearn.tree import DecisionTreeClassifier
features = [[120,0],[130,0],[80,1],[90,1],[85,1],[99,1],[140,0],[145,0]]
labels = ["apple","apple","orange","orange","orange","orange","apple","apple"]
```

```python
dlt = DecisionTreeClassifier()
dlt = dlt.fit(features,labels)
dlt.predict([[140,0]])   #output as apple
```

```python
import random
import time

for i in range(10):
  print(dlt.predict([[random.randint(60,150),random.randint(0,1)]]))
  time.sleep(2)
```
