Testing the Speed of the numpy

```python
#addition of list element using list
import time

size = 1_000_000_00 # at the value of 1_000_000_000 it will hang
l1 = list(range(size))
l2 = list(range(size))

start = time.time()
add = [x+y for x,y in zip(l1 ,l2)]
end = time.time()
print(end - start)
```
