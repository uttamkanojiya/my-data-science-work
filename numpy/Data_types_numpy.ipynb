Data Types in Numpy

Common Data Types In Numpy
1. int32 , int64
2. float32 , float64
3. bool
4. complex64 , complex128
5. Object

```python
#Finding out the data type in numpy
import numpy as np

arr = np.array([23,4,3,23])
print(arr)
print(arr.dtype)
```

```python
#Changing the data type in numpy
arr.astype("float32")
arr2 =np.array([23,4,3,23],dtype = np.float32)
arr3 = np.array([23,4,3,23],dtype = np.int32)
print(arr2)
```

Why Data Types Are Matters
1. Memory Usage
2. Performance
3. Precision

```python
#Finding how many bytes does it occupy
arr2.nbytes # this is of type float32
```

```python
arr3.nbytes # this is of type int32
```

```python
arr_int64 = np.array([1,2,3,],dtype = np.int64)
arr_int32 = np.array([1,2,3,],dtype = np.int32)
print("This is bytes occuied by int32 : ",arr_int32.nbytes)
print("This is bytes occuied by int64 : ",arr_int64.nbytes)
```

```python
#String data types in numpy
arr = np.array(['apple','banana',"cherry"],dtype="U10")
print(arr)
```

```python
#Complex numbers
arr = np.array([1 + 2j,3 + 4j,5 + 6j],dtype = "complex128")
print(arr)
```

```python
#Object Data type
arr = np.array([{'a':1},[1,2,3],"hello"],dtype = object)
print(arr)
```
