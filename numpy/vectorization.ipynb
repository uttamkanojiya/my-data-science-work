Vectorization 

```python
#Vectorization 
#squaring elements in python using list
import time

size = 1_000
list1 = list(range(size))
start = time.time()
list_squr = [x**2 for x in list1]
end = time.time()
print(end - start)
```

```python
#vectorization 
#squaring elements in numpy array
import numpy as np 
import time

size = 1_000
list1 = np.array(size)
start = time.time()
numpy_squares = list1**2
end = time.time()
print(end - start)
```
