BROADCASTING & VECTORIZATION

```python
import numpy as np

arr = np.array([1,2,3,4,5])
result = []
for i in arr:
    result.append(i**2)
print(result)
```

```python
#Another Method for same 
arr = np.array([1,2,3,4,5])
result = arr ** 2
print(result)
```

```python
arr = np.array([1,2,3,4,5])
result = arr + 10
print(result)
```

```python
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([1,2,3])
result = arr1 + arr2
print(result)
```

```python
#Normalizing Data using Braodcasting
data = np.array([[10,20,30],[15,25,35],[20,30,40],[25,35,45],[30,40,50]])
mean = data.mean(axis = 0)
std = data.std(axis = 0)
result = (data - mean) // std
print(result)
```
