DATA COLLECTION TECHNIQUE THROUGH WEB-SCRAPING

```python
from bs4 import BeautifulSoup
```

```python
items = []
for i in range(1,51):
    with open(f"html/page{i}.html","r",encoding="utf-8") as f:
        content = f.read()
    soup = BeautifulSoup(content,"html.parser")
    h3s = soup.find_all("h3")
    articles = soup.select("article.product_pod")
    for article in articles:
        title = article.find("h3").find("a")["title"]
        price = article.select_one("p.price_color").text.split("£")[1]
        rating_element = article.select_one("p.star-rating")
        rate = rating_element["class"][1]
        items.append([title,price,rate])
```

```python
import numpy as np 
shape = np.shape(items)
shape
```

```python
import pandas as pd
show = pd.DataFrame(items,columns=["Book Name","Prices","Ratings"])
show
```

```python
three = []
four = []
five = []
for item in items:
    if item[2] == "Three":
        three.append([item[0],item[2]])
    elif item[2] == "Four":
        four.append([item[0],item[2]])
    elif item[2] == "Five":
        five.append([item[0],item[2]])
    else:
        pass
print("Five star rating : ",five)
print("Four star rating : ",four)
print("Three star rating : ",three)
```

```python
final = show.to_csv("Data.csv")
```
