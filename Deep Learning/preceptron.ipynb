```python
def step(x):
    return 1 if x>= 0 else 0

def preceptron(x1,x2,w1,w2,b):
    y = x1*w1 + x2*w2 + b
    return step(y)

print(preceptron(0,0,1,1,-1.5))
print(preceptron(0,1,1,1,-1.5))
print(preceptron(1,0,1,1,-1.5))
print(preceptron(1,1,1,1,-1.5))

#We are try to get the logical AND gate values 
```
