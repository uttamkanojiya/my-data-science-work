```python
student = {
    "name" : "UTAM",
    "age" : 23,
    "grade" : "A"
}
student
```

```python
#Accessing the dictionary data
student["name"]
```

```python
# .get() method is also used in dictionary for accessing the data
student.get("name")
```

```python
#Updating the Dictionary existing value 
student["age"] = 67
```

```python
student
```

```python
student["name"] = "Uttam"
student
```

```python
#To display all the keys of the dictionary 
student.keys()
```

```python
# To display all the values in the dictionary
student.values()
```

```python
#To display all the key-value pair form dictionary
student.items()
```

```python
#To update or add more one key-value pair in dictionary 
student.update({"name" : "UTTAM"})
student
```

```python
student.update({"name" : "Robert", "language" : "Python"})
student
```

```python
grade = student.pop("grade") # Its removes the key and returns its value
grade
```

```python
student.popitem() # Removes and returns the last inserted key-value pair.
```

```python
student.setdefault("city" ,"Unknown") # Returns value for key, else sets it to default.
```

```python
student
```

```python
new_dict = student.copy() #returns a shallow copy of dictionary
```

```python
new_dict
```

```python
student.clear() ##It will delete all the items from the dictionary
```
