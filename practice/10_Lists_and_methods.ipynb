List and List Methods

```python
list1 = [3,5,23,3]
```

```python
list1[0] = 4    # changing the list item through index
list1
```

```python
list1.append(9)   #adding an new item at the end of list 
list1
```

```python
list2 = [6,7]
list1.extend(list2)  #it adds the new list to the old list at the end of the old list
list1
```

```python
list1.insert(1,200)    #its adds the new item at the particular index without removing any number 
list1
```

```python
nu = list1.pop(4)  #pop method will remove the item and also return the removed item 
nu
list1
```

```python
list1.index(4)  #return the place of item or index of item where it is located in list
```

```python
list1.index(23)
```

```python
list1.count(4)  #it counts the occurances of the item in the list
```

```python
list1.sort() #this will sort the list items in ascending order 
list1
```

```python
newl = [2,6,1,21,6]
sorted(newl)   #returns a new list which is sorted version of the original list
```
