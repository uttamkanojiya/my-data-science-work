```python
a = input("Enter the Value of a") # it takes string as input by default
print(a)
```

```python
b = input("Enter the Value of b") # it takes string as input by default
print(b)
```

```python
a + b
```

```python
int(a) + int(b) # typecasting of variables to from string to integer
```

```python
c = (int)(input("Enter the value of C : "))
```

```python
type(c)
```

```python
d = (str)(input("Enter the value for D : "))
type(d)
```

```python
e = (bool)(input("Enter the Value for E : ")) #it will takes anything as input then it show show True as output if nothing is given for input then output will be False
type(e)
```
