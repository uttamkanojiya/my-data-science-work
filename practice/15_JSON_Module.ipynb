JSON(JavaScript Object Notation) Module

```python
import json
obj = {"name" : "Uttam","Company" : "IBM","Salary" : 45000}
print(type(obj))
s = json.dumps(obj)      #This will convert the python object to JSON Strign
print(type(s))
print(s)
```

```python
d = json.loads('{"name": "Uttam", "Company": "IBM", "Salary": 45000}') #Converts the JSON string to python object
print(type(d))
print(d)
```

```python
#using dump()method to write json data into file 
file = open("data.json","w")
json.dump(obj,file)
file.close()
```

```python
#using the load() method to read from a json file
file = open("data.json","r")
data = json.load(file)
print(data)
file.close()
```

```python
# For Formatting the JSON Output 
formatted_json = json.dumps(data,indent = 4)
print(formatted_json)
```
