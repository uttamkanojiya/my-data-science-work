File Handling in Python

```python
f = open("text.txt","r") ##This will read the content from the specied file
f.read()
```

```python
f.close()
```

```python
s = "This is the best way to learn data science"
f = open("uttam.txt","w") #This will write the content in the specied file if not exists then it will create new and write
f.write(s)
f.close()
```

```python
r = "This is the best way to learn data science"
f = open("text.txt","w") #This will write the content in the specied file if not exists then it will create new and write
f.write(r)
f.close()
```

```python
#Using the with statment for file handling
with open("uttam.txt","r") as file:
    content = file.read()
    print(content) #No Need to manually close the file
```

```python
with open("text.txt","r") as file:
    content = file.read()
    print(content)
```

```python
text = "This is the New Text that is been added in text.txt"
with open("text.txt","w") as file:
    content = file.write(text)
    print(content) # When adding new text to existing file if its empty its okay but if its not then it will erase the existing data and write the new data
```
