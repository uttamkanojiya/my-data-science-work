OOP Concept (Object Oriented Programming) in python

```python
class Employee:
    company = "HP"
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    def printdetails(self):
        print(f"The Company of {self.name} is {self.company} with salary ${self.salary}")
e = Employee("UTTAM",54000)
e.printdetails()
```

```python
class Employee:
    company = "HP"
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    def printdetails(self):
        print(f"The Company of {self.name} is {self.company} with salary ${self.salary}")
    @staticmethod
    def printdata():
        print(f"Above is the Info of our Employee:")
e = Employee("UTTAM",54000)
e.printdetails()
e.printdata()
```

```python
class Employee:
    company = "HP"
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    def printdetails(self):
        print(f"The Company of {self.name} is {self.company} with salary ${self.salary}")
    @classmethod
    def printdata(cls):
        print(f"Printing values which are inside class:{cls.com}")
e = Employee("UTTAM",54000)
e.printdetails()
e.printdata()
```
