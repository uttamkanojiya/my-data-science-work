Loops In Python

```python
li = [3,4,5,67,78,89]
for item in li:
    print(item)
```

```python
for i in range(2,11):
    print(i)
```

```python
i = 0
while i < 12:
    print(i)
    i += 1
```

```python
for item in li:
    pass     #want to continue the loop statement later then use pass
print("Hello")
```

```python
li = ["apples","bananas",'tomatoes']
for item in li:
    if(item == "bananas"):
        break    # stop the execution right there
    print(item)
```

```python
for item in li:
    if (item == "bananas"):
        continue    # stop the iteration at particular condition and continue to next
    print(item)
```
