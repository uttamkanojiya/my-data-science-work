SET OPERATIONS 

```python
set1 = {1,2,3,4}
set2 = {3,4,5,6}
set1.union(set2) # it will give union of the sets and the resultunt set will be in new set
```

```python
sub = {1,3}
sub.issubset(set1) #Returns True if sub is a subset of set1.
```

```python
set1.issuperset(sub) # 	Returns True if set1 is a superset of sub.
```

```python
set1.difference(set2) # It return those element which is perent in set1 but not in set2
```

```python
set1.intersection(set2) # It return the element that is present in both the sets 
```

```python
set1.symmetric_difference(set2) # It return those element that is present either in set1 or set2 but not in both sets
```
