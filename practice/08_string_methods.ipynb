F-string and string formatting

```python
#Using format()
```

```python
name = "Uttam"
age = 100
print("My Name is {} and I am {} years old.".format(name,age))
```

```python
#Using f-string
```

```python
print(f"My Name is {name} and I am {age} years Old.")
```

```python
#formatting Numbers
```

```python
pi = 3.14159
print(f"Pi Rounded to 2 Decimal Places: {pi:.2f}")
```

```python
#Padding and  Formatting
```

```python
print(f"{'Uttam':<10}") #Its align the text to left
print(f"{'is':^10}") #Its align the text to center
print(f"{'Best':>10}") #Its align the text to right
```
