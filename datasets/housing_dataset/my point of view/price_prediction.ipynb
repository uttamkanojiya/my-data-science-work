```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import cross_val_score

#load the dataset
housing = pd.read_excel("final_dataset.xlsx")
```

```python
#housing[housing["Area"] == housing["Area"].max()] for knowing the maximum value
df = housing.iloc[:,1:-37]
housing["area_cat"] = pd.cut(
    housing["Area"],
    bins = [0,1000,2000,3000,5000,np.inf],
    labels = [1,2,3,4,5]
)
```

```python
#split the dataset into training and test
split = StratifiedShuffleSplit(n_splits=1,test_size=0.2,random_state=42)
for train_index,test_index in split.split(housing,housing["area_cat"]):
    strat_train_set = housing.loc[train_index].drop("area_cat",axis=1)
    strat_test_set = housing.loc[test_index].drop("area_cat",axis=1)
```

```python
#work with the copy of the training dataset
train_set_copy = strat_train_set.copy()
    #test_set_copy =strat_test_set.copy()  # just to see the count
    #test = test_set_copy.copy()
    #test.count

# separate the predictors and labels
labels = train_set_copy["Price"].copy()
features = train_set_copy.drop("Price",axis=1)
```

```python
#if there is any categorical and numerical data then separate it before giving it to model
```

```python
#Create the pipeline
num_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy = "median")),
    ("scaler",StandardScaler())
])
```

```python
#transform the data
housing_prepared = num_pipeline.fit_transform(train_set_copy)

#housing_prepared is now numpy array
print(housing_prepared.shape)
```

```python
#Train the model
#Taking various model for training like training to see which model is best suited

#Linear Regressior Model
lin_reg = LinearRegression()
lin_reg.fit(housing_prepared,labels)
lin_preds = lin_reg.predict(housing_prepared)
#lin_rmse = root_mean_squared_error(labels,lin_preds)
#print(f"The root mean squared error for linearRegression is : {lin_rmse}")
lin_rmses = -cross_val_score(lin_reg,housing_prepared,labels,scoring="neg_root_mean_squared_error",cv=10)
print(f"the root mean sqaured error using cross validation score for linear : {lin_rmses}")
print(pd.Series(lin_rmses).describe())
```

```python
#Decision Tree Model
dec_reg = DecisionTreeRegressor()
dec_reg.fit(housing_prepared,labels)
dec_preds = dec_reg.predict(housing_prepared)
dec_rmse = root_mean_squared_error(labels,dec_preds)
print(f"The root mean squared error for DecisionTreeRegressor : {dec_rmse}")
```

```python
#Random Forest Model
random_forest_reg = RandomForestRegressor()
random_forest_reg.fit(housing_prepared,labels)
random_forest_preds = random_forest_reg.predict(housing_prepared)
random_forest_rmse = root_mean_squared_error(labels,random_forest_preds)
print(f"The root mean sqaured error : {random_forest_rmse}")
```
