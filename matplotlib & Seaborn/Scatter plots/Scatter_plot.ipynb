SCATTER PLOTS IN MATPLOTLIB

```python
import matplotlib.pyplot as plt
import random 

study_hrs = [1,2,3,4,5,6,7,8,9]
exam_score = [45,24,54,52,25,57,23,12,90]
plt.scatter(study_hrs,exam_score)
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.grid(True)
plt.show()
```

```python
#Adding colors and szies
study_hrs = [1,2,3,4,5,6,7,8,9]
exam_score = [45,24,54,52,25,57,23,12,90]
colors = ['red' if score <= 35 else "green" for score in exam_score]
size = [score*5 for score in exam_score]
plt.scatter(study_hrs,exam_score,c=colors,s=size)
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.grid(True)
plt.show()
```

```python
#Using the color map
study_hrs = [1,2,3,4,5,6,7,8,9]
exam_score = [45,24,54,52,25,57,23,12,90]
plt.scatter(study_hrs,exam_score,c=exam_score,cmap="plasma")
plt.colorbar(label = "score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.grid(True)
plt.show()
```

```python
#using the Annotation 
study_hrs = [1,2,3,4,5,6,7,8,9]
exam_score = [45,24,54,52,25,57,23,12,90]
plt.scatter(study_hrs,exam_score,c=exam_score,cmap="plasma")
plt.colorbar(label="score")
for i in range(len(study_hrs)):
    plt.annotate(f'S{i+1}',(study_hrs[i]+0.2,exam_score[i]))
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.grid(True)
plt.show()
```

```python
#Creating multiple groups in one plot
class_a_hrs = [2,4,6,8]
class_a_scores = [44,55,66,88]
class_b_hrs = [1,3,5,7,9]
class_b_scores = [34,56,67,34,80]

plt.scatter(class_a_hrs,class_a_scores,label="Class A",color = "black")
plt.scatter(class_b_hrs,class_b_scores,label="Class B",color = "red")
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.title("Study Vs Marks")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
