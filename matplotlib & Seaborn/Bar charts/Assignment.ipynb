```python
import matplotlib.pyplot as plt
```

```python
import random 
import numpy as np
```

```python
kohil_runs = sorted(random.sample(range(500,2500),10))
sachin_runs = sorted(random.sample(range(100,2100),10))
sehwag_runs = sorted(random.sample(range(300,1900),10))

years = sorted(random.sample(range(1982,1992),10))
x = np.arange(len(years))
width = 0.25

plt.bar(x-width,kohil_runs,width=width,label="Kohils Runs")
plt.bar(x,sachin_runs,width=width,label="Sachins Runs")
plt.bar(x+width,sehwag_runs,width=width,label="Sehwag Runs")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Performance Comparison")
plt.legend()
plt.grid(True)
plt.xticks(x,years)
plt.show()
```

```python
players = ["Sachin","Yuvraj","Virat","Rohit"]
runs_5yrs = [300+400+1200+1500+1900,1400+0+0+900+1800,500+900+1200+1900+2100,900+1200+1500+900+1200]
plt.barh(players,runs_5yrs,color="skyblue")
plt.xlabel("Total runs in first 5 years")
plt.title("First 5 Years Performance Of Indian Batsman")
plt.tight_layout()
```
