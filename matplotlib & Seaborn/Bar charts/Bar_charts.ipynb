BAR CHARTS

```python
import matplotlib.pyplot as plt
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
plt.bar(years,runs)
plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Sachin Tendulkar Yearly Runs")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#Side-by-side bar charts
import numpy as np
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohil  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
x = np.arange(len(years))
width = 0.25

plt.bar(x - width,sachin, width=width,label="Sachin")
plt.bar(x,sehwag, width=width,label="Sehwag")
plt.bar(x + width,kohil, width=width,label="Kohil")
plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Performance Comparison")
plt.xticks(x,years)
plt.legend()
plt.show()
```

```python
#Horizontal Bar charts
players = ["Sachin","Sehwag","Kohil","Yuvraj"]
run_5yrs = [500+700+1100+1500+1800,0+200+900+1400+1600,0+0+500+800+1100,300+600+800+1100+900]
plt.barh(players,run_5yrs,color="skyblue")
plt.xlabel("Total Runs in First 5 years")
plt.title("First 5-years performance of indian batsman")
plt.tight_layout()
plt.show()
```

```python
#Adding value labels with plt.text()
players = ["Sachin","Sehwag","Kohil"]
runs = [1500,1200,1800]
plt.bar(players,runs,color="skyblue")
for i in range(len(players)):
    plt.text(i,runs[i]+20,str(runs[i]),ha="center")
plt.ylabel("Runs")
plt.title("Runs scored by player")
plt.tight_layout()
plt.show()
```
