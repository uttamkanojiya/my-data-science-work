BAR CHARTS

```python
import matplotlib.pyplot as plt
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
plt.bar(years,runs)
plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Sachin Tendulkar Yearly Runs")
plt.grid(True)
plt.tight_layout()
plt.show()
```
