HISTOGRAM IN MATPLOTLIB

```python
import matplotlib.pyplot as plt
import random

data = sorted(random.sample(range(10,500),300))
plt.hist(data,bins = 10,edgecolor="white",label="Marks in exam")
plt.xlabel("MARKS")
plt.ylabel("FREQUENCY")
plt.grid(True)
plt.legend()
plt.style.use("Solarize_Light2")
plt.tight_layout()
```

```python
ages = sorted(random.sample(range(10,80),50))
bins = [10,20,30,40,50,60,70]
plt.hist(ages,bins=bins,label = "Youtube Viewers",edgecolor="white",color="black")
plt.xlabel("Different Age Groups")
plt.ylabel("Frequency of Viewers")
plt.grid(True)
plt.tight_layout()
plt.legend()
plt.title("Different Age Groups Watching Youtube")
plt.show()
```

```python
numbers = [random.choice(range(10, 71, 7)) for _ in range(100)]
bins = [10,20,30,40,50,60,70]
plt.hist(numbers,bins=bins,color="black",edgecolor="white",label = "Some data")
plt.xlabel = ("Numbers")
plt.ylabel = ("frequency")
plt.title("some data")
plt.axvline(40,color="red",linestyle="--",linewidth=2,label="Average")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.style.use("ggplot")
plt.show()
```
