DATA VISUALIZATION USING MATPLOTLIB & SEABORN

```python
import matplotlib.pyplot as plt

plt.plot([12,23,34],[56,67,75]) # this type gragh representation is not recommended
plt.show()
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
plt.plot(years,runs) # Also this type gragh is also not recommended bcz no proper label
plt.show()
```

```python
#Showing multiple lines in one gragh
sachine =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
kohil = [0,0,500,800,1100,1300,1500,1800,1900,2100]
sehwag = [0,300,800,1200,1500,1700,1600,1400,1000,0]

plt.plot(years,sachine,"rv--",label="Sachine Tendulkars",linewidth=4)
plt.plot(years,kohil,label="Virat Kohil",linewidth=4)
plt.plot(years,sehwag,label="Virendra Sehwag",linewidth=4)
plt.legend()
plt.tight_layout()
plt.style.use("dark_background")
plt.grid(True)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Performance Comparsion")
plt.show()
```

```python
#Using format strings
plt.plot(years,kohil,"ro--",label="kohil")
plt.plot(years,sachine,"yo--",label="Sachin")
plt.plot(years,sehwag,"g^:",label="sehwag")
plt.grid(True)
plt.legend()
#plt.plot(years,runs,color="blue",label="Tendulkars")
```

```python
#Virualixing the tons of data - What crowded looks link
import numpy as np
for i in range(50):
    plt.plot(np.random.rand(100),linewidth=1)
plt.title("Too Much Data can be confusing")
plt.grid(True)
plt.tight_layout()
plt.show()
```
