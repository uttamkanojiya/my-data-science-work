STACK PLOT

```python
import matplotlib.pyplot as plt
days = [1,2,3,4,5,6,7]
studying = [3,4,3,5,4,3,4]
playing = [2,2,1,1,2,3,2]
watching_tv = [2,1,2,2,1,1,1]
sleeping = [5,5,6,5,6,5,5]
labels = ["Studying","PLaying","Watching TV","Sleeping"]
colors = ["blue",'green','red','yellow']
plt.figure(figsize=(6,6))
plt.stackplot(days,studying,playing,watching_tv,sleeping,labels=labels,colors=colors,alpha=0.3) #alpha is used to control the opacity
plt.legend(loc="upper right")
plt.xlabel("Days")
plt.ylabel("Hours")
plt.title("Your Daily Hours Spent")
plt.grid(True)
plt.tight_layout()
plt.show()
```
