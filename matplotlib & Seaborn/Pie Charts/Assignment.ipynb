      Create a pie chart showing the market share of mobile OS(Android ,iOS,others).Then Create the same         using a horizontal bar chart and observe which is easier to understand

```python
import matplotlib.pyplot as plt

mobile_os = ["Android","iOS","Others.."]
usage = [72,28,6]
explode = [0.2,0,0.4]
colors= ["#008BFF","#E4FF30","#FA8112"]
plt.pie(usage,labels=mobile_os,explode=explode,colors=colors,shadow=True,autopct="%1.1f%%",startangle=120)
plt.show()
```
