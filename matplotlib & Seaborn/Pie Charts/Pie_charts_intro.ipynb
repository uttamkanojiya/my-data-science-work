PIE CHARTS

```python
import matplotlib.pyplot as plt

#Basic Pie charts example
labels = ["Sachin","Sehwag","Kohil","Yuvraj"]
runs = [18000,8000,12000,9500]
plt.title("Career Runs Of Indian Batsmen")
plt.pie(runs,labels=labels)
plt.show()
```

```python
#For Custom colors
colors = ["#7C00FE","#F9E400","#FFAF00","#F5004F"]
plt.pie(runs,labels=labels,colors=colors)
plt.show()
```

```python
#Adding the edges and style slices with wedgepros
plt.pie(runs,labels=labels,colors=colors,wedgeprops={'edgecolor':"black","linewidth":2,'linestyle':"--","fill":True})
```

```python
#Highlight a slice using explode
explode = [0.1,0,0,0]
plt.pie(runs,labels=labels,explode=explode,colors=colors)
plt.title("Exploded Pie Example")
plt.show()
```

```python
#Add Shadows for a 3D feel
plt.pie(runs,labels=labels,explode=explode,colors=colors,shadow=True)
plt.show()
```

```python
#Start Angle
plt.pie(runs,labels=labels,explode=explode,colors=colors,shadow=True,startangle=140)
plt.show()
```

```python
#Show Percentages with autopct
plt.pie(runs,labels=labels,autopct="%1.1f%%",startangle=140)
plt.show()
```

```python
#Crowded Pie Chart Example(Not Recommended)
languages = ["Python","Java",'C++',"JavaScript",'C#',"Ruby",'Go',"Rust",'Swift','PHP']
usage = [30,20,10,10,7,5,4,3,2,1]
plt.pie(usage,labels=languages,autopct="%1.1f%%",startangle=90)
plt.title("Programming Languages Usgae (Crowded Example)")
plt.show()
```
