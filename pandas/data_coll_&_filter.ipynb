```python
#selecting multiple data or fields from the given excel file
data[["List Price","Actual Price"]]
```

```python
#Selecting rows by index (method1)
#It will give the complete row by label
data.loc[4]
```

```python
#Selecting rows by index (method2)
#It will give the complete row by position
data.iloc[4]
```

```python
#Selecting particular rows and columns 
data.loc[4,'Store']
```

```python
data.loc[3,'Store']
```

```python
#By .iloc method
data.iloc[4,8]
```
