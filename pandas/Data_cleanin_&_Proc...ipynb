DATA CLEANING AND PROCESSING 

```python
import pandas as pd

#firstly check weather the data is containing some null values are not
data = pd.read_csv("data_sample.csv")
data.isnull()
```

```python
#Counting the Missing Value per column
data.isnull().sum()
```

```python
#to drop the nul value rows
df = data.copy()
df
```

```python
df.dropna()
```

```python
data
```

```python
#want to fill the null data with some values
df.fillna(0)
```

```python
#Replacing the null value to its columns average
df["Age"].fillna(df["Age"].mean())
```

```python
df
```

```python
#replacing the null value through help of forward and backward fill
df.ffill()
```

```python
df.bfill()
```

```python
#detecting and removing the duplicates values from the data
df.duplicated()
```

```python
df
```

```python
#to remove the duplicated value from the data
df.drop_duplicates()
```

```python
#Checking for duplicate values from multiple columns
df.duplicated(subset=["Name","Age"])
```
