Merging & Joins

```python
import pandas as pd
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
inner_join = pd.merge(employees,departments,on="DeptID") #By default merging is done on the basis of inner join
a=""" this is an inner join where the two dataframes are added on basis of common data where both dataframes
    common rows and columns will be seen."""
inner_join
```

```python
left_join = pd.merge(employees,departments,on="DeptID",how="left")
b = """this is an left join where the two dataframes are merge on the basis of 
left table or dataframe given, Here the left dataframes is employees. 
"""
left_join
```

```python
right_join = pd.merge(employees,departments,on="DeptID",how="right")
c = """this is an right join where the two dataframes are added on the basis of right dataframe given 
here the right dataframe is departments.
"""
right_join
```

```python
outer_join = pd.merge(employees,departments,on = "DeptID",how="outer")
d = """this is an outer join examplewhere the two dataframes are added on the basis of both the dataframes 
here both dataframes are added if any rows or columns doesnt have values then its given NaN there.
"""
outer_join
```

```python
#Concating the Dataframes
df1 = pd.DataFrame({"Name": ["Alice", "Bob"]})
df2 = pd.DataFrame({"Name": ["Charlie", "David"]})

pd.concat([df1, df2]) #Concating dataframes in vertical way
```

```python
pd.concat([df1, df2],axis=1) #concating dataframes in horizontal way
```
