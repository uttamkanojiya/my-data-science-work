Melt & Pivot

```python
data = {
    "Name" : ["Alice","Bob","Charlie"],
    "Math" : [85,78,92],
    "science" : [90,45,75],
    "English" : [45,68,45]
}
```

```python
import pandas as pd

data = pd.DataFrame(data)
data
```

```python
#Using the Melt()
df1 = data.copy()
df1.melt(id_vars=['Name'],value_vars=["Math","science","English"],var_name="Subject",value_name="score")
```

```python
#Using the pivot method
df1.pivot(index="Name",columns="Subject",values="score")
```
