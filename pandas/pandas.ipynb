```python
#Creating a series data type in pandas
import pandas as pd

df = pd.Series([124,1324,3141,34,134,1414])
print(type(df))
print(df)
```

```python
#Providing are own index
df1 = pd.Series([124,1324,3141,34,134,1414],index = ["Ramesh","jayesh",'deepak','mangesh','rohit','sharad'])
print(df1)
```

```python
#Creating dataframe
df1 = pd.Series([124,1324,3141,34,134,1414],index = ["Ramesh","jayesh",'deepak','mangesh','rohit','sharad'])
show = pd.DataFrame(df1)
print(show)
```

```python
data = {
    "name" : ['Sandesh','Akash',"Avinash"],
    "age" : [34,45,56],
    "city" : ["Delhi","bangalore","dombivli"]
}
df = pd.DataFrame(data)
print(df)
```

```python
data = [["pooja",32],["deepti",23],["siri",344]]
df = pd.DataFrame(data,columns =["name",'age'])
print(df)
```
