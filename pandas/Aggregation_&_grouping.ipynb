Aggregation & Grouping

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  
df
```

```python
#to know the average salary of particular department
df.groupby("Department")["Salary"].mean()
```

```python
#Common Aggregation
mean = df.groupby("Team")["Salary"].mean()
print("Mean Salary of the Team is : ",mean)
print()
sum_all = df.groupby("Team")["Salary"].sum()
print("Sum of the Salary of all the team is : ",sum_all)
print()
count  = df.groupby("Team")["Salary"].count()
print("Count of the team Salary is : ",count)
print()
max_all = df.groupby("Team")["Salary"].max()
print("Max of the Team Salary is : ",max_all)
print()
min_all = df.groupby("Team")["Salary"].min()
print("Min of the team salary is : ",min_all)
```

```python
#Grouping multiple columns by groupby 
multi = df.groupby(["Team","Gender"])["Salary"].mean()
multi
```

```python
#Custom Aggregation with .agg()
df.groupby("Team")["Salary"].agg(["mean","max","min"])
```

```python
#name your own function
df.groupby("Team")["Salary"].aggregate(avg_score = "mean",high_score="max")
```

```python
#Also can be done by the .agg() method bcz both .agg() and .aggregate() function are exactly same
df.groupby("Team")["Salary"].agg(avg_score = "mean",high_score="max")
```

```python
#apply differnet functions to different columns 
df.groupby("Team").agg({'Salary':"mean",'Age':"max"})
```

```python
#Using the transform method
df["Team_avg"] = df.groupby("Team")["Salary"].transform("mean")
```

```python
#Using the .filter method
df.groupby("Team").filter(lambda x: x["Salary"].mean()>85)
```

```python
#Using the Aggregate function
df.groupby("Team").aggregate({"Salary":"mean","Age":"max"})
```
